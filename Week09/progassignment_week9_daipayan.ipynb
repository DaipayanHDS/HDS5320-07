```python
import time

result_scores = {}

def Create_Linear_Models(solve):
    clf = LogisticRegression(fit_intercept=True, penalty=None, solver=solve, max_iter=10000)

    time1 = time.time()
    clf.fit(X_train, y_train)
    time2 = time.time()

    result_scores[f"{solve}"] = (
        accuracy_score(y_train, clf.predict(X_train)),
        accuracy_score(y_test, clf.predict(X_test)),
        time2 - time1
    )

def get_result():
    print("\n{0:20}   {1:25}   {2:25}   {3:10}".format(
        'Solver used', 'Training subset accuracy', 'Test subset accuracy', 'Time taken'
    ))
    print('---------------------------------------------------------------------------------------------')
    for solver, (train_acc, test_acc, time_taken) in result_scores.items():
        print("{0:20}   {1:<25.4f}   {2:<25.4f}   {3:<10.4f}".format(
            solver, train_acc, test_acc, time_taken
        ))

Create_Linear_Models("lbfgs")
Create_Linear_Models("newton-cg")
Create_Linear_Models("newton-cholesky")
Create_Linear_Models("sag")
Create_Linear_Models("saga")
```
