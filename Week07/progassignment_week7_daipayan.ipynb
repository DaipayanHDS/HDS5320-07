
## Spark Application

Every year, CMS publishes complete datasets that consolidate the information submitted by reporting entities for active years of Open Payments data.

In this exercise, we are going to look at Research payment dataset from CMS 2023 Open Payments file. This dataset contains records of payments or transfer of value made in connection with formal research agreements or protocols. This includes funding for clinical trials and other research activities.

<cite>CMS 2023 Open Payments : https://download.cms.gov/openpayments/PGYR2023_P01302025_01212025.zip</cite>

```
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, BooleanType, DateType
import pyspark.sql.functions

# File location and type
file_location = "/FileStore/tables/OP_DTL_RSRCH_PGYR2023_P01302025_01212025.csv"
file_type = "csv"

# CSV options
infer_schema = "true"
first_row_is_header = "true"
delimiter = ","

# The applied options are for CSV files. For other file types, these will be ignored.
rsrch_df = spark.read.format(file_type) \
  .option("inferSchema", infer_schema) \
  .option("header", first_row_is_header) \
  .option("sep", delimiter) \
  .load(file_location)

display(rsrch_df)
```

The following is Covered Recipient (Physician) Profile Dataset, it provides demographic information about physicians, including specialties and license details. Each physician is assigned a unique identification number (Physician Profile ID) to facilitate data searches across other datasets.

<cite>CMS 2023 Covered Recipient Profile Dataset : https://download.cms.gov/openpayments/PHPRFL_P01302025_01212025.zip</cite>

```
# File location and type
file_location = "/FileStore/tables/OP_CVRD_RCPNT_PRFL_SPLMTL_P01302025_01212025.csv"
file_type = "csv"

# CSV options
infer_schema = "true"
first_row_is_header = "true"
delimiter = ","

# The applied options are for CSV files. For other file types, these will be ignored.
cvrd_rec_df = spark.read.format(file_type) \
  .option("inferSchema", infer_schema) \
  .option("header", first_row_is_header) \
  .option("sep", delimiter) \
  .load(file_location)

display(cvrd_rec_df)
```

### 1. What is the Nature of Payments with reimbursement amounts greater than $1,000 ordered by count?

```
rsrch_df = rsrch_df.withColumn("Total_Amount_of_Payment_USDollars",rsrch_df.Total_Amount_of_Payment_USDollars.cast(IntegerType()))

payment_df = rsrch_df.selectExpr("Total_Amount_of_Payment_USDollars", "Form_of_Payment_or_Transfer_of_Value") \
  .where("Total_Amount_of_Payment_USDollars > 1000") \
  .groupBy("Form_of_Payment_or_Transfer_of_Value") \
  .count()
```

```
payment_df.display()
```

### 2. What are the top ten Nature of Payments by count?

```
from pyspark.sql.functions import desc
mop_count_df = rsrch_df.na.drop(subset=["Form_of_Payment_or_Transfer_of_Value"]) \
    .groupBy("Form_of_Payment_or_Transfer_of_Value") \
    .count() \
    .orderBy(desc("count"))
```

```
mop_count_df.display()
```

### 3. What are the top ten Nature of Payments by total amount?

```
from pyspark.sql.functions import sum
mop_amount_df = rsrch_df.na.drop(subset=["Form_of_Payment_or_Transfer_of_Value"]) \
    .groupBy("Form_of_Payment_or_Transfer_of_Value") \
    .agg(sum("Total_Amount_of_Payment_USDollars").alias("Total amount($)")) \
    .orderBy(desc('Total amount($)'))
```

```
mop_amount_df.display()
```

### 4. What are the top ten physician specialties by total amount?

```
speciality_df = rsrch_df.na.drop(subset=["Principal_Investigator_1_Specialty_1"]) \
    .groupBy("Principal_Investigator_1_Specialty_1") \
    .agg(sum("Total_Amount_of_Payment_USDollars").alias("Total amount($)")) \
    .orderBy(desc("Total amount($)")) \
    .limit(10)
```

```
speciality_df.display()
```

### 5. Who are the top ten physicians by total amount?

```
phy_df = rsrch_df.selectExpr("Principal_Investigator_1_Profile_ID", "Principal_Investigator_1_First_Name", "Principal_Investigator_1_Last_Name") \
    .distinct()
```

```
phy_amount_df = rsrch_df.na.drop(subset=["Principal_Investigator_1_Profile_ID","Total_Amount_of_Payment_USDollars"]) \
    .groupBy("Principal_Investigator_1_Profile_ID") \
    .agg(sum("Total_Amount_of_Payment_USDollars").alias("Total amount($)")) \
    .limit(10)
phy_amount_df = phy_amount_df.join(phy_df, on="Principal_Investigator_1_Profile_ID", how="inner") \
    .orderBy(desc('Total amount($)'))
```

```
phy_amount_df.display() 
```
